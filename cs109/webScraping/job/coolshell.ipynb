```python
%load_ext autoreload
%autoreload 2

def readers_reverse(s):
    fields = s.split(",")
    return int(fields[2])

parts = ["0_2", "3_7", "8_18", "18_38", "38_58", "58_78"]
list_of_articles = []

for part in parts:
    file_name= "data/coolshell/coolshell" + part + ".txt"
    with open(file_name, "r", encoding="utf-8") as f:
        articles = f.readlines()
        list_of_articles.extend(articles)

list_of_articles.sort(key=readers_reverse, reverse=True)

for a in list_of_articles:
    file_name= "data/coolshell/sorted_all.txt"
    with open(file_name, "w", encoding="utf-8") as f:
        f.writelines(list_of_articles)
```

```python
%load_ext autoreload
%autoreload 2

import time

from coolshell import *

ts = "20251202192447"
url="https://web.archive.org/web/"+ ts+ "/https://coolshell.cn/"

cs = coolshell([], [])

results = []
offset = 58
step = 20

for i in range(step):
    j = i + offset
    if j < 74:
        time.sleep(3)
        pn = "" if j == 0 else "page/" + str(j + 1)
        urlp = url + pn
        results.extend(cs.process_page(urlp))

with open("coolshell" + str(offset) + "_" + str(offset+step) + ".txt", "w", encoding="utf-8") as f:
    f.writelines(results)
```
